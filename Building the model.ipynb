```python
import os
import json
from datetime import datetime
from typing import Dict, Any, List

import numpy as np
import pandas as pd
from sklearn.preprocessing import MultiLabelBinarizer
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    f1_score, precision_score, recall_score, roc_auc_score
)
import joblib

import torch
from torch.utils.data import Dataset

from transformers import (
    AutoTokenizer,
    AutoConfig,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer,
    DataCollatorWithPadding,
    EarlyStoppingCallback
)

# -----------------------
# USER SETTINGS / PATHS
# -----------------------
MODEL_NAME = "roberta-base"
SAVE_DIR = "/content/drive/MyDrive/Moniepoint/model"
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

os.makedirs(SAVE_DIR, exist_ok=True)

# ---------- Data prep ----------
# Make a defensive copy
df = output_df.copy()

# Convert 'at' to datetime (user requested). If format ambiguous, infer
df['at'] = pd.to_datetime(df['at'], errors='coerce', infer_datetime_format=True)

# Drop rows with invalid dates or missing content
df = df.dropna(subset=['at', 'content', 'Category']).reset_index(drop=True)

# Split into list-of-labels column by semicolon separator (and strip)
def split_labels(cell: str) -> List[str]:
    # Some cells may already be lists, or have separators like ';' or ','; adapt as needed
    if isinstance(cell, list):
        labels = cell
    else:
        labels = [s.strip() for s in str(cell).split(';') if s.strip() != ""]
    return labels

df['labels_list'] = df['Category'].map(split_labels)

# Define date cutoff: any row on or after 2025-09-01 is test
cutoff = pd.to_datetime("2025-09-01")

train_df = df[df['at'] < cutoff].reset_index(drop=True)
test_df  = df[df['at'] >= cutoff].reset_index(drop=True)

print(f"Total rows: {len(df)}  |  Train: {len(train_df)}  |  Test: {len(test_df)}")

# ---------- Label binarization ----------
mlb = MultiLabelBinarizer(sparse_output=False)
y_train = mlb.fit_transform(train_df['labels_list'])
y_test  = mlb.transform(test_df['labels_list'])

labels = list(mlb.classes_)
num_labels = len(labels)
print(f"Found {num_labels} labels: {labels}")

# Save labels mapping (human-readable)
with open(os.path.join(SAVE_DIR, "labels.json"), "w", encoding="utf-8") as f:
    json.dump(labels, f, ensure_ascii=False, indent=2)

# Save mlb (so you can transform at inference time offline)
joblib.dump(mlb, os.path.join(SAVE_DIR, "mlb.joblib"))

# ---------- Tokenizer and Config ----------
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME, use_fast=True)
config = AutoConfig.from_pretrained(
    MODEL_NAME,
    num_labels=num_labels,
    problem_type="multi_label_classification"
)
model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME, config=config)
model.to(DEVICE)

# ---------- Dataset wrapper ----------
class ReviewsDataset(Dataset):
    def __init__(self, texts: List[str], labels: np.ndarray, tokenizer, max_length: int = 256):
        self.texts = texts
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_length = max_length

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        txt = str(self.texts[idx])
        encoded = self.tokenizer(
            txt,
            truncation=True,
            max_length=self.max_length,
            return_attention_mask=True
        )
        item = {k: torch.tensor(v) for k, v in encoded.items()}
        item["labels"] = torch.tensor(self.labels[idx], dtype=torch.float)  # float for BCEWithLogitsLoss
        return item

train_dataset = ReviewsDataset(train_df['content'].tolist(), y_train, tokenizer)
eval_dataset  = ReviewsDataset(test_df['content'].tolist(), y_test, tokenizer)

data_collator = DataCollatorWithPadding(tokenizer=tokenizer)

# ---------- Metrics ----------
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def compute_metrics(eval_pred) -> Dict[str, Any]:
    logits, labels = eval_pred
    probs = sigmoid(logits)
    # choose threshold 0.5 (you can tune this later)
    preds = (probs >= 0.5).astype(int)

    metrics = {}
    # Micro / Macro F1, precision, recall
    try:
        metrics['f1_micro'] = f1_score(labels, preds, average='micro', zero_division=0)
        metrics['f1_macro'] = f1_score(labels, preds, average='macro', zero_division=0)
        metrics['precision_micro'] = precision_score(labels, preds, average='micro', zero_division=0)
        metrics['recall_micro'] = recall_score(labels, preds, average='micro', zero_division=0)
    except Exception as e:
        print("Error computing F1/precision/recall:", e)

    # roc_auc_score per label (only if there is variation in y_true for the label)
    try:
        # roc_auc_score requires at least one positive and one negative for each label
        valid_label_mask = [len(np.unique(labels[:, i])) > 1 for i in range(labels.shape[1])]
        if any(valid_label_mask):
            metrics['roc_auc_micro'] = roc_auc_score(labels[:, valid_label_mask].ravel(),
                                                    probs[:, valid_label_mask].ravel())
        else:
            metrics['roc_auc_micro'] = None
    except Exception as e:
        metrics['roc_auc_micro'] = None
        print("ROC AUC computation error:", e)

    return metrics

# ---------- Training arguments ----------
# Fixed parameter names for older transformers versions
training_args = TrainingArguments(
    output_dir=SAVE_DIR,
    eval_strategy="epoch",
    save_strategy="epoch",
    logging_strategy="steps",
    logging_steps=200,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=16,
    num_train_epochs=3,
    weight_decay=0.01,
    learning_rate=2e-5,
    load_best_model_at_end=True,
    metric_for_best_model="f1_micro",
    greater_is_better=True,
    save_total_limit=3,
    fp16=torch.cuda.is_available(),
)

# ---------- Trainer ----------
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=eval_dataset,
    tokenizer=tokenizer,
    data_collator=data_collator,
    compute_metrics=compute_metrics,
    callbacks=[EarlyStoppingCallback(early_stopping_patience=2)]
)

# ---------- Start training ----------
trainer.train()

# ---------- Final evaluation ----------
print("Evaluating on test set...")
eval_metrics = trainer.evaluate(eval_dataset=eval_dataset)
print(eval_metrics)

# ---------- Save everything for offline use ----------
print("Saving model and tokenizer to:", SAVE_DIR)
model.save_pretrained(SAVE_DIR)
tokenizer.save_pretrained(SAVE_DIR)

# mlb already saved. Also save a small metadata file
meta = {
    "model_name_or_path": MODEL_NAME,
    "num_labels": num_labels,
    "labels": labels,
    "cutoff_date_for_test": "2025-09-01",
    "saved_at": datetime.utcnow().isoformat() + "Z"
}
with open(os.path.join(SAVE_DIR, "metadata.json"), "w", encoding="utf-8") as f:
    json.dump(meta, f, ensure_ascii=False, indent=2)

print("All artifacts saved. Components in:", SAVE_DIR)
```

```python
import os
import json
import joblib
import torch
import numpy as np
from transformers import AutoTokenizer, AutoModelForSequenceClassification

# -----------------------
# Load artifacts
# -----------------------
SAVE_DIR = "./model"

# Load model & tokenizer
tokenizer = AutoTokenizer.from_pretrained(SAVE_DIR)
model = AutoModelForSequenceClassification.from_pretrained(SAVE_DIR)
model.eval()

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
model.to(DEVICE)

# Load MultiLabelBinarizer and labels
mlb = joblib.load(os.path.join(SAVE_DIR, "mlb.joblib"))
with open(os.path.join(SAVE_DIR, "labels.json"), "r", encoding="utf-8") as f:
    labels = json.load(f)

# Sigmoid for probabilities
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# -----------------------
# Fixed Prediction function
# -----------------------
def predict(texts, threshold=0.5):
    """Predict categories for a list of texts."""
    if isinstance(texts, str):
        texts = [texts]

    # Tokenize
    encodings = tokenizer(
        texts,
        truncation=True,
        padding=True,
        max_length=256,
        return_tensors="pt"
    ).to(DEVICE)

    # Forward pass
    with torch.no_grad():
        outputs = model(**encodings)
        logits = outputs.logits.cpu().numpy()

    # Convert to probabilities
    probs = sigmoid(logits)

    # Apply threshold
    pred_bin = (probs >= threshold).astype(int)

    # Decode to label names - FIX: Convert to numpy array first
    all_preds = []
    for row in pred_bin:
        # Ensure row is 2D array for inverse_transform
        row_2d = np.array([row])
        categories = mlb.inverse_transform(row_2d)[0]
        all_preds.append(categories)

    return all_preds, probs


sample_texts = [
    "The app crashes every time I try to open it.",
    "Please add a dark mode and reduce charges."
]

predictions, probabilities = predict(sample_texts, threshold=0.5)

for txt, pred, prob in zip(sample_texts, predictions, probabilities):
    print(f"\nText: {txt}")
    print(f"Predicted categories: {pred}")
```
