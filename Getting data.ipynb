```python
from google_play_scraper import app, Sort, reviews_all
from app_store_scraper import AppStore
import pandas as pd
import numpy as np
import json, os, uuid

import warnings

# Suppress all warnings
warnings.filterwarnings('ignore')
```

```python
gg_reviews = reviews_all(
    "com.moniepoint.personal",
    sleep_milliseconds=0,
    lang='en', # defaults to 'en'
    country='ng', # defaults to 'us'
    sort=Sort.NEWEST, # defaults to Sort.NEWEST
)
```

```python
df = pd.DataFrame(gg_reviews)
df.head()
```

```python
import os

# Keep only reviewId and content
data = df[['reviewId', 'content']]

# Target directory
output_dir = r"C:\Users\HP\Desktop\Multi-Labelled-Review-Categorization-Model\d"
os.makedirs(output_dir, exist_ok=True)

# Split dataframe into chunks of 55 rows
chunk_size = 55
for i in range(0, len(data), chunk_size):
    chunk = data.iloc[i:i+chunk_size]
    
    # Define file name
    file_name = f"reviews_part_{i//chunk_size + 1}.txt"
    file_path = os.path.join(output_dir, file_name)
    
    # Save as txt (tab-separated or space-separated)
    chunk.to_csv(file_path, sep="\t", index=False, header=True, encoding="utf-8")
    
print("Done! Files saved in:", output_dir)
```
